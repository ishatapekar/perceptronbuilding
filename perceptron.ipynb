```python
import tensorflow as tf
import numpy as np
```

```python
## Students data: [hours studied, hours slept]

X = np.array([[2, 3], [1, 2], [3, 4], [5, 6], [4, 5], [3, 4]], dtype=np.float32)

## Label: 0 = Fail, 1 = Pass

y = np.array([[1], [0], [1], [1], [0], [1]], dtype=np.float32)

print("Input data ", X)
print("Labels: ", y)
```

tf.keras follows a consistent machine learning workflow:

1.Define the Model: Arrange layers to create a neural network.
2.Compile: Specify the optimizer, loss function, and metrics.
3.Fit: Train the model on data.
4.Evaluate: Test performance on unseen data.
5.Predict: Use the model to generate outputs for new inputs.

```python
## Model creation/definition

model = tf.keras.Sequential(
    [tf.keras.layers.Dense(units = 1,input_shape = (2,),activation = 'sigmoid')]  
)
```

```python
model.summary()
```

```python
model.compile(
    optimizer = 'sgd',
    loss = 'binary_crossentropy',
    metrics = ['accuracy']
)
```

```python
## Model training

model.fit(X, y, epochs=100, verbose=1)
```

```python
# Evaluate model

loss, accuracy = model.evaluate(X, y, verbose=1)
```

```python
# Make predictions 

new_data = np.array([[4,4], [8,5], [6,9]], dtype=float)
predictions = model.predict(new_data)

print(predictions)
```

```python
for i in range(len(new_data)):
    print(f"Student with hours studied: {new_data[i][0]} and hours slept: {new_data[i][1]}")

    prob1 = predictions[i][0]

    result = "Passed" if prob1 > 0.5 else "Failed"
    print(f"Prediction: {prob1}, Result: {result}")
```
